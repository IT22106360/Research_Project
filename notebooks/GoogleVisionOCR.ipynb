```python
import os
from PIL import Image

SRC_DIR = r"D:\Y4 Research\datasets\new_dieatary\subset1_1"
OUT_DIR = r"D:\Y4 Research\datasets\new_dieatary\subset1_1_resized"

TARGET_W = 2048
TARGET_H = 1536

os.makedirs(OUT_DIR, exist_ok=True)

for filename in os.listdir(SRC_DIR):
    if filename.lower().endswith((".png", ".jpg", ".jpeg", ".tiff")):
        src_path = os.path.join(SRC_DIR, filename)
        out_path = os.path.join(OUT_DIR, filename)

        try:
            with Image.open(src_path).convert("RGB") as img:
                w, h = img.size

                scale = min(TARGET_W / w, TARGET_H / h)
                new_w = int(w * scale)
                new_h = int(h * scale)

                resized = img.resize((new_w, new_h), Image.LANCZOS)

                canvas = Image.new("RGB", (TARGET_W, TARGET_H), "white")
                x_offset = (TARGET_W - new_w) // 2
                y_offset = (TARGET_H - new_h) // 2

                canvas.paste(resized, (x_offset, y_offset))
                canvas.save(out_path, quality=95)

                print(f"✅ {filename}: {w}×{h} → {TARGET_W}×{TARGET_H}")

        except Exception as e:
            print(f"⚠️ {filename}: error → {e}")
```

```python
import os
from PIL import Image

IMAGE_DIR = r"D:\Y4 Research\datasets\new_dieatary\subset1_1_resized"
REQUIRED_DPI = 300

for filename in os.listdir(IMAGE_DIR):
    if filename.lower().endswith((".png", ".jpg", ".jpeg", ".tiff")):
        path = os.path.join(IMAGE_DIR, filename)

        try:
            with Image.open(path) as img:
                dpi = img.info.get("dpi", None)

                if dpi is None:
                    print(f"⚠️ {filename}: DPI metadata missing")
                else:
                    x_dpi, y_dpi = dpi

                    if int(x_dpi) == REQUIRED_DPI and int(y_dpi) == REQUIRED_DPI:
                        print(f"✅ {filename}: {x_dpi}×{y_dpi} DPI")
                    else:
                        print(f"❌ {filename}: {x_dpi}×{y_dpi} DPI")

        except Exception as e:
            print(f"⚠️ {filename}: error → {e}")
```

```python
from google.cloud import vision
import io
import json
import os

# ---------------- OCR FUNCTION ----------------
def detect_text_with_bboxes(path, image_id):
    client = vision.ImageAnnotatorClient()

    with io.open(path, "rb") as image_file:
        content = image_file.read()

    image = vision.Image(content=content)
    response = client.document_text_detection(image=image)

    if response.error.message:
        raise Exception(response.error.message)

    annotation = response.full_text_annotation

    full_text = annotation.text
    words_output = []

    for page in annotation.pages:
        for block in page.blocks:
            for paragraph in block.paragraphs:
                for word in paragraph.words:

                    word_text = "".join(symbol.text for symbol in word.symbols)

                    confidences = [symbol.confidence for symbol in word.symbols]
                    avg_conf = int(sum(confidences) / len(confidences) * 100)

                    vertices = word.bounding_box.vertices
                    x_coords = [v.x for v in vertices]
                    y_coords = [v.y for v in vertices]

                    x_min = min(x_coords)
                    y_min = min(y_coords)
                    width = max(x_coords) - x_min
                    height = max(y_coords) - y_min

                    words_output.append({
                        "text": word_text,
                        "confidence": avg_conf,
                        "bbox": [x_min, y_min, width, height]
                    })

    return {
        "image_id": image_id,
        "ocr_text": full_text,
        "words": words_output
    }


# ---------------- BATCH PROCESS ----------------
input_dir = r"D:\Y4 Research\datasets\new_dieatary\subset1_1_resized"
output_dir = r"D:\Y4 Research\datasets\new_dieatary\subset1_1_ocr_jsons"

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Supported image formats
image_extensions = (".png", ".jpg", ".jpeg", ".tiff", ".bmp")

for filename in os.listdir(input_dir):
    if filename.lower().endswith(image_extensions):

        image_path = os.path.join(input_dir, filename)
        image_id = os.path.splitext(filename)[0]  # e.g. 622.png → 622
        output_path = os.path.join(output_dir, f"{image_id}.json")

        try:
            print(f"Processing: {filename}")
            result = detect_text_with_bboxes(image_path, image_id)

            with open(output_path, "w", encoding="utf-8") as f:
                json.dump(result, f, indent=4, ensure_ascii=False)

            print(f"Saved → {output_path}\n")

        except Exception as e:
            print(f"❌ Failed on {filename}: {e}\n")

print("✅ Batch OCR processing completed.")
```
