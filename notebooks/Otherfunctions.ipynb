```python
import os

def remove_json_extension_from_folders(base_dir):
    """
    Removes '.json' extension from folder names inside base_dir.
    Only affects immediate subfolders.
    """
    for name in os.listdir(base_dir):
        old_path = os.path.join(base_dir, name)

        # Only folders ending with .json
        if os.path.isdir(old_path) and name.lower().endswith(".json"):
            new_name = name[:-5]  # remove '.json'
            new_path = os.path.join(base_dir, new_name)

            if os.path.exists(new_path):
                print(f"⚠️ Skipped (target exists): {new_name}")
                continue

            os.rename(old_path, new_path)
            print(f"✅ Renamed: {name} → {new_name}")
```

```python
import os

def align_file_name_with_folder(base_dir):
    """
    For each folder in base_dir:
    - If the file inside does not have the same name as the folder,
      rename the file to match the folder name.
    - Skips folders with 0 or >1 files to avoid ambiguity.
    """

    for folder_name in os.listdir(base_dir):
        folder_path = os.path.join(base_dir, folder_name)

        if not os.path.isdir(folder_path):
            continue

        files = [
            f for f in os.listdir(folder_path)
            if os.path.isfile(os.path.join(folder_path, f))
        ]

        if len(files) == 0:
            print(f"⚠️ Empty folder: {folder_name}")
            continue

        if len(files) > 1:
            print(f"⚠️ Multiple files in folder, skipped: {folder_name}")
            continue

        file_name = files[0]
        file_path = os.path.join(folder_path, file_name)

        file_base, file_ext = os.path.splitext(file_name)

        # If names already match, skip
        if file_base == folder_name:
            continue

        new_file_name = folder_name + file_ext
        new_file_path = os.path.join(folder_path, new_file_name)

        if os.path.exists(new_file_path):
            print(f"⚠️ Target exists, skipped: {new_file_name}")
            continue

        os.rename(file_path, new_file_path)
        print(f"✅ Renamed: {file_name} → {new_file_name}")
```

```python
from paddleocr import PaddleOCR
import json
from PIL import Image

def extract_ocr_data(image_path, image_id):
    """
    Extract OCR data from an image and format it as tokens and bounding boxes.
    
    Args:
        image_path: Path to the image file
        image_id: Identifier for the image
    
    Returns:
        Dictionary with image_id, tokens, and bboxes
    """
    # Initialize PaddleOCR
    ocr = PaddleOCR(use_textline_orientation=True, lang='en')
    
    # Perform OCR
    result = ocr.predict(image_path)
    
    # Initialize output structure
    output = {
        "image_id": image_id,
        "tokens": [],
        "bboxes": []
    }
    
    # Debug: print the result structure to understand the format
    print("Result type:", type(result))
    if isinstance(result, dict):
        print("Dict keys:", result.keys())
        if result:
            print("Sample keys and types:", {k: type(v) for k, v in list(result.items())[:3]})
    elif isinstance(result, list) and len(result) > 0:
        print("List length:", len(result))
        print("First item type:", type(result[0]))
        if isinstance(result[0], list) and len(result[0]) > 0:
            print("Sample item:", result[0][0])
    
    # Process OCR results - handle new PaddleOCR format
    if result:
        # Check if result is a dict (new format) or list (old format)
        if isinstance(result, dict):
            # New format: result is a dict with 'dt_polys' and 'rec_text'
            if 'dt_polys' in result and 'rec_text' in result:
                bboxes = result['dt_polys']
                texts = result['rec_text']
                
                for bbox, text_info in zip(bboxes, texts):
                    # Extract text (might be a tuple or string)
                    text = text_info[0] if isinstance(text_info, tuple) else text_info
                    
                    # Convert bbox to [x1, y1, x2, y2]
                    if len(bbox) == 4 and len(bbox[0]) == 2:  # [[x1,y1],[x2,y2],[x3,y3],[x4,y4]]
                        x_coords = [point[0] for point in bbox]
                        y_coords = [point[1] for point in bbox]
                    elif len(bbox) == 4:  # [x1, y1, x2, y2]
                        x1, y1, x2, y2 = bbox
                        x_coords = [x1, x2]
                        y_coords = [y1, y2]
                    else:
                        continue
                    
                    x1 = int(min(x_coords))
                    y1 = int(min(y_coords))
                    x2 = int(max(x_coords))
                    y2 = int(max(y_coords))
                    
                    # Split text into tokens
                    words = text.split()
                    
                    if len(words) > 1:
                        word_width = (x2 - x1) / len(words)
                        for i, word in enumerate(words):
                            word_x1 = int(x1 + i * word_width)
                            word_x2 = int(x1 + (i + 1) * word_width)
                            
                            output["tokens"].append(word)
                            output["bboxes"].append([word_x1, y1, word_x2, y2])
                    else:
                        output["tokens"].append(text)
                        output["bboxes"].append([x1, y1, x2, y2])
        
        # Old format: result is a list
        elif isinstance(result, list) and len(result) > 0 and result[0]:
            for line in result[0]:
                print(f"Processing line: {line}")  # Debug
                print(f"Line type: {type(line)}, Length: {len(line) if hasattr(line, '__len__') else 'N/A'}")  # Debug
                
                # Extract bounding box coordinates and text
                bbox = line[0]
                text = line[1]
                
                print(f"Bbox: {bbox}, type: {type(bbox)}")  # Debug
                print(f"Text: {text}, type: {type(text)}")  # Debug
                
                # Handle both old format (text, confidence) and new format (just text)
                if isinstance(text, tuple):
                    text, confidence = text
                
                # Convert bbox - handle different formats
                if isinstance(bbox, (list, tuple)) and len(bbox) > 0:
                    # Check if it's [[x1,y1],[x2,y2],...] format
                    if isinstance(bbox[0], (list, tuple)) and len(bbox[0]) == 2:
                        x_coords = [point[0] for point in bbox]
                        y_coords = [point[1] for point in bbox]
                    # Check if it's [x1, y1, x2, y2] format
                    elif len(bbox) == 4 and all(isinstance(x, (int, float)) for x in bbox):
                        x_coords = [bbox[0], bbox[2]]
                        y_coords = [bbox[1], bbox[3]]
                    else:
                        print(f"Unknown bbox format: {bbox}")
                        continue
                else:
                    print(f"Invalid bbox: {bbox}")
                    continue
                
                x1 = int(min(x_coords))
                y1 = int(min(y_coords))
                x2 = int(max(x_coords))
                y2 = int(max(y_coords))
                
                # Split text into tokens (words)
                words = text.split()
                
                # For multi-word lines, distribute bbox proportionally
                if len(words) > 1:
                    word_width = (x2 - x1) / len(words)
                    for i, word in enumerate(words):
                        word_x1 = int(x1 + i * word_width)
                        word_x2 = int(x1 + (i + 1) * word_width)
                        
                        output["tokens"].append(word)
                        output["bboxes"].append([word_x1, y1, word_x2, y2])
                else:
                    # Single word
                    output["tokens"].append(text)
                    output["bboxes"].append([x1, y1, x2, y2])
    
    return output

# Example usage
if __name__ == "__main__":
    # Process your vitamin B2 label image
    image_path = r"D:\Y4 Research\datasets\dietary Images\Set1\542 conv 0.png"
    image_id = "542 conv 0"
    
    # Extract OCR data
    result = extract_ocr_data(image_path, image_id)
    
    # Print formatted JSON
    print(json.dumps(result, indent=2))
    
    # Optionally save to file
    with open(r"D:\Y4 Research\datasets\dietary Images\annotation.json", "w") as f:
        json.dump(result, f, indent=2)
    
    print(f"\nExtracted {len(result['tokens'])} tokens")
```

```python
from paddleocr import PaddleOCR
import json

def extract_ocr_data(image_path, image_id):
    """
    Extract OCR data from an image and format it with detailed word-level information.
    
    Args:
        image_path: Path to the image file
        image_id: Identifier for the image
    
    Returns:
        Dictionary with image_id, ocr_text, and words array
    """
    # Initialize PaddleOCR
    ocr = PaddleOCR(use_textline_orientation=True, lang='en')
    
    # Perform OCR
    result = ocr.predict(image_path)
    
    # Initialize output structure
    output = {
        "image_id": image_id,
        "ocr_text": "",
        "words": []
    }
    
    all_text = []
    
    # Handle dict format (newer PaddleOCR)
    if isinstance(result, dict):
        if 'dt_polys' in result and 'rec_text' in result:
            bboxes = result['dt_polys']
            texts = result['rec_text']
            
            for bbox, text_info in zip(bboxes, texts):
                # Extract text and confidence
                if isinstance(text_info, (list, tuple)) and len(text_info) >= 2:
                    text = text_info[0]
                    confidence = int(text_info[1] * 100)  # Convert to percentage
                else:
                    text = str(text_info)
                    confidence = 96  # Default confidence
                
                all_text.append(text)
                
                # Convert bbox to [x, y, width, height]
                if isinstance(bbox, (list, tuple)) and len(bbox) >= 4:
                    if isinstance(bbox[0], (list, tuple)):
                        # Format: [[x1,y1],[x2,y2],[x3,y3],[x4,y4]]
                        x_coords = [point[0] for point in bbox]
                        y_coords = [point[1] for point in bbox]
                        x = int(min(x_coords))
                        y = int(min(y_coords))
                        width = int(max(x_coords) - x)
                        height = int(max(y_coords) - y)
                    else:
                        # Format: [x1, y1, x2, y2]
                        x = int(bbox[0])
                        y = int(bbox[1])
                        width = int(bbox[2] - bbox[0])
                        height = int(bbox[3] - bbox[1])
                    
                    output["words"].append({
                        "text": text,
                        "confidence": confidence,
                        "bbox": [x, y, width, height]
                    })
    
    # Handle list format (older PaddleOCR)
    elif isinstance(result, list) and len(result) > 0:
        for page in result:
            if not page:
                continue
            for line in page:
                if len(line) < 2:
                    continue
                
                bbox = line[0]
                text_info = line[1]
                
                # Extract text and confidence
                if isinstance(text_info, (list, tuple)) and len(text_info) >= 2:
                    text = text_info[0]
                    confidence = int(text_info[1] * 100)
                else:
                    text = str(text_info)
                    confidence = 96
                
                all_text.append(text)
                
                # Convert bbox
                if isinstance(bbox, (list, tuple)) and len(bbox) >= 4:
                    if isinstance(bbox[0], (list, tuple)) and len(bbox[0]) >= 2:
                        # Format: [[x1,y1],[x2,y2],[x3,y3],[x4,y4]]
                        x_coords = [point[0] for point in bbox]
                        y_coords = [point[1] for point in bbox]
                        x = int(min(x_coords))
                        y = int(min(y_coords))
                        width = int(max(x_coords) - x)
                        height = int(max(y_coords) - y)
                    elif all(isinstance(coord, (int, float)) for coord in bbox):
                        # Format: [x1, y1, x2, y2]
                        x = int(bbox[0])
                        y = int(bbox[1])
                        width = int(bbox[2] - bbox[0])
                        height = int(bbox[3] - bbox[1])
                    else:
                        continue
                    
                    output["words"].append({
                        "text": text,
                        "confidence": confidence,
                        "bbox": [x, y, width, height]
                    })
    
    # Combine all text into ocr_text field
    output["ocr_text"] = " ".join(all_text)
    
    return output

# Example usage
if __name__ == "__main__":
    # Process your image
    image_path = r"D:\Y4 Research\datasets\dietary Images\Set1\542 conv 0.png"
    image_id = "000_000_022_2303_4"
    
    # Extract OCR data
    result = extract_ocr_data(image_path, image_id)
    
    # Print formatted JSON
    print(json.dumps(result, indent=2))
    
    # Optionally save to file
    with open(r"E:\annotation.json", "w", encoding="utf-8") as f:
        json.dump(result, f, indent=2, ensure_ascii=False)
    
    print(f"\nExtracted {len(result['words'])} words")
    print(f"Total OCR text length: {len(result['ocr_text'])} characters")
```

```python
import json
import cv2
import matplotlib.pyplot as plt

# -------------------------------
# PATHS (YOUR FILES)
# -------------------------------
IMG_PATH = r"D:\Y4 Research\datasets\dietary Images\Set1\542 conv 0.png"
JSON_PATH = r"D:\Y4 Research\datasets\dietary Images\sample_annotation.json"

# OCR image size used when OCR was generated
OCR_W = 2480
OCR_H = 3508

# -------------------------------
# LOAD IMAGE
# -------------------------------
image = cv2.imread(IMG_PATH)
if image is None:
    raise FileNotFoundError(f"Cannot load image: {IMG_PATH}")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
H, W, _ = image.shape

print("Image size:", W, H)

# -------------------------------
# LOAD OCR JSON
# -------------------------------
with open(JSON_PATH, "r", encoding="utf-8") as f:
    ocr = json.load(f)

tokens = ocr["tokens"]
bboxes = ocr["bboxes"]

# -------------------------------
# HELPERS
# -------------------------------
def rescale_box(box, src_w, src_h, dst_w, dst_h):
    x1, y1, x2, y2 = box
    return [
        int(x1 / src_w * dst_w),
        int(y1 / src_h * dst_h),
        int(x2 / src_w * dst_w),
        int(y2 / src_h * dst_h),
    ]

def clamp_pixel_box(box, width, height):
    x1, y1, x2, y2 = box
    return [
        max(0, min(x1, width)),
        max(0, min(y1, height)),
        max(0, min(x2, width)),
        max(0, min(y2, height)),
    ]

def normalize_box(box, width, height):
    x1, y1, x2, y2 = box
    return [
        int(1000 * x1 / width),
        int(1000 * y1 / height),
        int(1000 * x2 / width),
        int(1000 * y2 / height),
    ]

def clamp_norm_box(box):
    return [max(0, min(v, 1000)) for v in box]

# -------------------------------
# FIX PIPELINE (CANONICAL)
# -------------------------------
def fix_bboxes(bboxes, image, ocr_size):
    H, W, _ = image.shape
    ow, oh = ocr_size

    # 1️⃣ Rescale OCR boxes → current image size
    bboxes_px = [rescale_box(b, ow, oh, W, H) for b in bboxes]

    # 2️⃣ Clamp pixel boxes
    bboxes_px = [clamp_pixel_box(b, W, H) for b in bboxes_px]

    # 3️⃣ Validate pixel boxes
    for i, b in enumerate(bboxes_px):
        if not (b[0] < b[2] and b[1] < b[3]):
            raise ValueError(f"Invalid pixel box at index {i}: {b}")

    # 4️⃣ Normalize for LayoutLMv3
    bboxes_norm = [normalize_box(b, W, H) for b in bboxes_px]
    bboxes_norm = [clamp_norm_box(b) for b in bboxes_norm]

    # 5️⃣ Validate normalized boxes
    for i, b in enumerate(bboxes_norm):
        if not all(0 <= v <= 1000 for v in b):
            raise ValueError(f"Invalid normalized box at index {i}: {b}")

    return bboxes_px, bboxes_norm

# -------------------------------
# APPLY FIX
# -------------------------------
bboxes_px, bboxes_norm = fix_bboxes(
    bboxes=bboxes,
    image=image,
    ocr_size=(OCR_W, OCR_H)
)

print("Bounding boxes fixed successfully")
print("Pixel boxes:", len(bboxes_px))
print("Normalized boxes:", len(bboxes_norm))

# -------------------------------
# VISUALIZE (PIXEL BOXES)
# -------------------------------
for token, box in zip(tokens, bboxes_px):
    x1, y1, x2, y2 = box
    cv2.rectangle(image, (x1, y1), (x2, y2), (255, 0, 0), 2)

plt.figure(figsize=(10, 6))
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
import os
import json
import re

# Folder containing JSON files
json_dir = r"D:\Y4 Research\All_PNGS_dieatary\subset1_1_jsons"

for filename in os.listdir(json_dir):
    if not filename.lower().endswith(".json"):
        continue

    old_path = os.path.join(json_dir, filename)

    try:
        with open(old_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        # Handle both dict and list cases
        if isinstance(data, list):
            image_path = data[0]["image"]
        else:
            image_path = data["image"]

        # Extract number before .png
        match = re.search(r"(\d+)\.png$", image_path)
        if not match:
            print(f"❌ No number found in image path: {image_path}")
            continue

        image_number = match.group(1)
        new_filename = f"{image_number}.json"
        new_path = os.path.join(json_dir, new_filename)

        # Rename file
        os.rename(old_path, new_path)
        print(f"✅ Renamed: {filename} → {new_filename}")

    except Exception as e:
        print(f"⚠️ Error processing {filename}: {e}")
```

```python
from google.cloud import vision
import io
import json


def detect_text_with_bboxes(path, image_id):
    client = vision.ImageAnnotatorClient()

    with io.open(path, "rb") as image_file:
        content = image_file.read()

    image = vision.Image(content=content)
    response = client.document_text_detection(image=image)

    if response.error.message:
        raise Exception(response.error.message)

    annotation = response.full_text_annotation

    # Full OCR text
    full_text = annotation.text

    words_output = []

    for page in annotation.pages:
        for block in page.blocks:
            for paragraph in block.paragraphs:
                for word in paragraph.words:
                    word_text = "".join(
                        [symbol.text for symbol in word.symbols]
                    )

                    # Average symbol confidence → word confidence
                    confidences = [symbol.confidence for symbol in word.symbols]
                    avg_conf = int(sum(confidences) / len(confidences) * 100)

                    # Bounding box
                    vertices = word.bounding_box.vertices
                    x_coords = [v.x for v in vertices]
                    y_coords = [v.y for v in vertices]

                    x_min = min(x_coords)
                    y_min = min(y_coords)
                    width = max(x_coords) - x_min
                    height = max(y_coords) - y_min

                    words_output.append({
                        "text": word_text,
                        "confidence": avg_conf,
                        "bbox": [x_min, y_min, width, height]
                    })

    result = {
        "image_id": image_id,
        "ocr_text": full_text,
        "words": words_output
    }

    return result


# ---------------- RUN ----------------
path_to = r"D:\Y4 Research\All_PNGS_dieatary\subset1_1\542.png"
image_id = "test"

output = detect_text_with_bboxes(path_to, image_id)

print(json.dumps(output, indent=4))
```
