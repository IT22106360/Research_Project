```python
import os
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# Set the path to your images directory
image_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\raw"

# Get all PNG files from the directory
image_files = [f for f in os.listdir(image_dir) if f.endswith('.jpg')]

print(f"Found {len(image_files)} PNG images in the directory")

# Display basic information about the dataset
if image_files:
    print(f"\nFirst few image names:")
    for img_name in image_files[:5]:
        print(f"  - {img_name}")

# Function to display a grid of images
def display_images(image_dir, image_files, num_images=9, figsize=(15, 15)):
    """
    Display a grid of images from the dataset
    
    Parameters:
    - image_dir: directory path
    - image_files: list of image filenames
    - num_images: number of images to display (default: 9)
    - figsize: figure size for the plot
    """
    num_images = min(num_images, len(image_files))
    cols = 3
    rows = (num_images + cols - 1) // cols
    
    fig, axes = plt.subplots(rows, cols, figsize=figsize)
    axes = axes.flatten() if num_images > 1 else [axes]
    
    for i in range(num_images):
        img_path = os.path.join(image_dir, image_files[i])
        img = Image.open(img_path)
        
        axes[i].imshow(img)
        axes[i].set_title(image_files[i], fontsize=10)
        axes[i].axis('off')
    
    # Hide any unused subplots
    for i in range(num_images, len(axes)):
        axes[i].axis('off')
    
    plt.tight_layout()
    plt.show()

# Display the first 9 images
if image_files:
    display_images(image_dir, image_files, num_images=9)
```

```python

# Source folder
src_folder = r"D:\Y4 Research\datasets\ingredient & nutrition Images\raw"

# Destination folder for PNGs
dst_folder = r"D:\Y4 Research\datasets\ingredient & nutrition Images\png_converted"
os.makedirs(dst_folder, exist_ok=True)

# Loop through all JPG files and convert to PNG
for filename in os.listdir(src_folder):
    if filename.lower().endswith(".jpg"):
        jpg_path = os.path.join(src_folder, filename)
        png_filename = os.path.splitext(filename)[0] + ".png"
        png_path = os.path.join(dst_folder, png_filename)

        # Open JPG and save as PNG
        with Image.open(jpg_path) as img:
            img.save(png_path, "PNG")

        print(f"Converted: {filename} -> {png_filename}")

print("All JPG images have been converted to PNG in the folder 'png_converted'.")
```

```python
# -----------------------------
# PATHS (LOCAL MACHINE)
# -----------------------------
input_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\png_converted"
output_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\resized_1800_dpi300"

os.makedirs(output_dir, exist_ok=True)

TARGET_WIDTH = 1800
processed_images = []

# -----------------------------
# RESIZE + SAVE WITH 300 DPI
# -----------------------------
for filename in sorted(os.listdir(input_dir)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        img = cv2.imread(input_path)
        if img is None:
            print(f"Skipping unreadable image: {filename}")
            continue

        h, w = img.shape[:2]

        # Upscale only if width < 1800
        if w < TARGET_WIDTH:
            scale = TARGET_WIDTH / w
            new_w = TARGET_WIDTH
            new_h = int(h * scale)

            img = cv2.resize(
                img,
                (new_w, new_h),
                interpolation=cv2.INTER_CUBIC
            )

        # Save temporarily with OpenCV
        cv2.imwrite(output_path, img)

        # Re-save with PIL to set DPI metadata
        pil_img = Image.open(output_path)
        pil_img.save(output_path, dpi=(300, 300))

        processed_images.append(output_path)

print(f"Total processed images: {len(processed_images)}")

# -----------------------------
# DISPLAY FIRST 6 IMAGES
# -----------------------------
num_to_show = min(6, len(processed_images))

for i in range(num_to_show):
    img = cv2.imread(processed_images[i])
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(6, 8))
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"Processed Image {i+1}")

plt.show()
```

```python
#only normalization illumination
# -----------------------------
# PATHS
# -----------------------------
input_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\resized_1800_dpi300"
output_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_only"

os.makedirs(output_dir, exist_ok=True)

processed_images = []

# -----------------------------
# PROCESS IMAGES
# -----------------------------
for filename in sorted(os.listdir(input_dir)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        img = cv2.imread(input_path)
        if img is None:
            print(f"Skipping unreadable image: {filename}")
            continue

        # 1. Convert to 8-bit grayscale
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

        # 2. Apply CLAHE only (no brightness standardization)
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        gray_clahe = clahe.apply(gray)

        # Save result
        cv2.imwrite(output_path, gray_clahe)
        processed_images.append(output_path)

print(f"✅Processed {len(processed_images)} images with CLAHE only.")

# -----------------------------
# DISPLAY 6 RANDOM IMAGES
# -----------------------------
num_to_show = min(6, len(processed_images))
sample_images = random.sample(processed_images, num_to_show)

for i, img_path in enumerate(sample_images, start=1):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(6, 8))
    plt.imshow(img, cmap="gray")
    plt.axis("off")
    plt.title(f"CLAHE Only ({i})")

plt.show()
```

```python
#Denoising
input_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_only"
output_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_denoised"

os.makedirs(output_dir, exist_ok=True)

processed_images = []

# -----------------------------
# NON-LOCAL MEANS DENOISING PARAMETERS
# -----------------------------
h = 8                   # Filter strength. Increase for more denoising
templateWindowSize = 7  # Size of template patch
searchWindowSize = 21   # Size of search window

# -----------------------------
# PROCESS IMAGES
# -----------------------------
for filename in sorted(os.listdir(input_dir)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        img = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            print(f"Skipping unreadable image: {filename}")
            continue

        # Apply Non-Local Means Denoising
        denoised = cv2.fastNlMeansDenoising(
            img,
            None,
            h=h,
            templateWindowSize=templateWindowSize,
            searchWindowSize=searchWindowSize
        )

        # Save result
        cv2.imwrite(output_path, denoised)
        processed_images.append(output_path)

print(f"✅ Processed {len(processed_images)} images with Non-Local Means denoising.")

# -----------------------------
# DISPLAY 6 RANDOM DENOISED IMAGES
# -----------------------------
num_to_show = min(6, len(processed_images))
sample_images = random.sample(processed_images, num_to_show)

for i, img_path in enumerate(sample_images, start=1):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(6, 8))
    plt.imshow(img, cmap="gray")
    plt.axis("off")
    plt.title(f"CLAHE + Denoised ({i})")

plt.show()
```

```python
#Contrast enhancement

# -----------------------------
# PATHS
# -----------------------------
input_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_denoised"
output_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_denoised_enhanced"

os.makedirs(output_dir, exist_ok=True)

processed_images = []

# -----------------------------
# CLAHE PARAMETERS
# -----------------------------
clip_limit = 2.5
tile_grid_size = (12, 12)

clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)

# -----------------------------
# PROCESS IMAGES
# -----------------------------
for filename in sorted(os.listdir(input_dir)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        img = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            print(f"Skipping unreadable image: {filename}")
            continue

        # Apply CLAHE contrast enhancement
        enhanced = clahe.apply(img)

        # Save result
        cv2.imwrite(output_path, enhanced)
        processed_images.append(output_path)

print(f"✅ Processed {len(processed_images)} images with CLAHE contrast enhancement.")

# -----------------------------
# DISPLAY 6 RANDOM ENHANCED IMAGES
# -----------------------------
num_to_show = min(6, len(processed_images))
sample_images = random.sample(processed_images, num_to_show)

for i, img_path in enumerate(sample_images, start=1):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(6, 8))
    plt.imshow(img, cmap="gray")
    plt.axis("off")
    plt.title(f"CLAHE Enhanced ({i})")

plt.show()
```

```python
# Deskewing

# -----------------------------
# PATHS
# -----------------------------
input_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_denoised_enhanced"
output_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\deskewed"

os.makedirs(output_dir, exist_ok=True)

processed_images = []

# -----------------------------
# FUNCTION TO CALCULATE SKEW ANGLE
# -----------------------------
def compute_skew(image):
    # Invert colors: text = white, background = black
    img_bin = cv2.bitwise_not(image)
    
    # Threshold to get binary image
    _, img_bin = cv2.threshold(img_bin, 0, 255, cv2.THRESH_BINARY | cv2.THRESH_OTSU)
    
    coords = np.column_stack(np.where(img_bin > 0))
    angle = cv2.minAreaRect(coords)[-1]
    
    if angle < -45:
        angle = -(90 + angle)
    else:
        angle = -angle
    return angle

# -----------------------------
# DESKEWING FUNCTION
# -----------------------------
def deskew_image(image, angle):
    (h, w) = image.shape
    center = (w // 2, h // 2)
    M = cv2.getRotationMatrix2D(center, angle, 1.0)
    deskewed = cv2.warpAffine(image, M, (w, h), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)
    return deskewed

# -----------------------------
# PROCESS IMAGES
# -----------------------------
for filename in sorted(os.listdir(input_dir)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        img = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            print(f"Skipping unreadable image: {filename}")
            continue

        angle = compute_skew(img)
        deskewed = deskew_image(img, angle)

        cv2.imwrite(output_path, deskewed)
        processed_images.append(output_path)

print(f"✅ Processed {len(processed_images)} images with skew correction.")

# -----------------------------
# DISPLAY 6 RANDOM DESKEWED IMAGES
# -----------------------------
num_to_show = min(6, len(processed_images))
sample_images = random.sample(processed_images, num_to_show)

for i, img_path in enumerate(sample_images, start=1):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(6, 8))
    plt.imshow(img, cmap="gray")
    plt.axis("off")
    plt.title(f"Deskewed Image ({i})")

plt.show()
```

```python
#Final Image Sharpening 

# -----------------------------
# PATHS
# -----------------------------
input_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\grayscale_clahe_denoised_enhanced"
output_dir = r"D:\Y4 Research\datasets\ingredient & nutrition Images\sharpened"

os.makedirs(output_dir, exist_ok=True)

processed_images = []

# -----------------------------
# SHARPENING KERNEL
# -----------------------------
# Unsharp mask approach
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

# -----------------------------
# PROCESS IMAGES
# -----------------------------
for filename in sorted(os.listdir(input_dir)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        img = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            print(f"Skipping unreadable image: {filename}")
            continue

        # Apply sharpening
        sharpened = cv2.filter2D(img, -1, kernel)

        # Save result
        cv2.imwrite(output_path, sharpened)
        processed_images.append(output_path)

print(f"✅ Processed {len(processed_images)} images with final sharpening.")

# -----------------------------
# DISPLAY 6 RANDOM SHARPENED IMAGES
# -----------------------------
num_to_show = min(6, len(processed_images))
sample_images = random.sample(processed_images, num_to_show)

for i, img_path in enumerate(sample_images, start=1):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(6, 8))
    plt.imshow(img, cmap="gray")
    plt.axis("off")
    plt.title(f"Sharpened Image ({i})")

plt.show()
```
